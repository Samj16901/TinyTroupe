# Creating and Validating Agents

This notebook demonstrates the process of creating `TinyPerson` agents with specific personas using `TinyPersonFactory` and then validating their characteristics against a set of defined expectations using `TinyPersonValidator`.

We will:
1. Define a context and particular specifications for different agent archetypes (e.g., a Banker, a Monk).
2. Use `TinyPersonFactory` to generate these agents based on the provided descriptions.
3. Inspect the generated agent's basic information (minibio).
4. Define a set of textual expectations for each agent archetype.
5. Use `TinyPersonValidator` to assess how well the generated agent aligns with these expectations, receiving a score and a justification.
6. Showcase how validation works by also checking an agent against a mismatched set of expectations.

## 1. Setup and Imports

Import necessary classes from TinyTroupe: `TinyPerson` for the agent representation, `TinyPersonFactory` for generating agents, and `TinyPersonValidator` for checking their characteristics. We also import `textwrap` for better display of long text outputs.

```python
import json
import sys
# If running from the 'examples' directory, this adds the parent directory (project root) to the Python path
sys.path.insert(0, '..')

import tinytroupe # Initializes configuration, logging, etc.
from tinytroupe.agent import TinyPerson
# TinyWorld and TinySocialNetwork are not strictly needed for this example but often used with agents.
# from tinytroupe.environment import TinyWorld, TinySocialNetwork 
from tinytroupe.factory import TinyPersonFactory
from tinytroupe.validation import TinyPersonValidator
# ResultsReducer and control are not used in this specific notebook's flow.
# from tinytroupe.extraction import ResultsReducer
# import tinytroupe.control as control

import textwrap
```

## 2. Creating a "Banker" Agent

First, we define the general context for the factory (the environment or organization the agent belongs to) and the specific particularities of the agent we want to create.

### 2.1. Define Specifications for the Banker

```python
# General context for the factory: the type of bank
bank_context_spec =\
"""
A large Brazilian bank. It has many branches and a large number of employees. It is currently facing significant competition from fintechs and is focused on digital transformation and innovation while maintaining its strong corporate client base.
"""

# Specific particularities for the banker agent we want to generate
banker_particularities_spec =\
"""
A dynamic and experienced Vice-President at one of Brazil's largest banks. This executive has a background in engineering and holds an MBA in finance. They are known for their strategic thinking and are deeply involved in the bank's efforts to innovate and compete with emerging fintech companies. Must be fluent in Portuguese and English.
"""
```

### 2.2. Generate the Banker Agent

We create a `TinyPersonFactory` instance with the bank context and then use it to generate the banker agent with its specific particularities.

```python
# Create a factory for generating people within the bank's context
banker_factory = TinyPersonFactory(context_text=bank_context_spec)

# Generate the banker agent
banker = banker_factory.generate_person(agent_particularities=banker_particularities_spec)

if banker:
    print(f"Successfully generated agent: {banker.name}")
else:
    print("Agent generation failed. Check logs for details.")
```

### 2.3. Inspect the Generated Agent's Minibio

The `minibio()` method gives a concise summary of the agent's generated persona.

```python
if banker:
    print(textwrap.fill(banker.minibio(), width=100))
```

### 2.4. Define Validation Expectations for the Banker

To validate the generated agent, we define a set of expectations. These are textual descriptions of the traits, tastes, and characteristics the agent *should* possess based on our initial specifications. Keeping these expectations separate from the generation prompt helps in obtaining an unbiased evaluation.

```python
banker_expectations =\
"""
The individual should be:
 - Affluent and financially successful, consistent with a Vice-President role at a major bank.
 - Highly intelligent, ambitious, and a strategic thinker.
 - Well-connected within the finance industry and possibly other sectors.
 - Likely in their 40s or 50s, given their experience and role.

Regarding tastes and lifestyle:
  - Enjoys international travel.
  - Possesses cultivated hobbies such as reading widely, collecting art, or playing golf.
  - Appreciates high-quality experiences, including fine wines and gourmet food.
  - May express concerns about excessive taxation or overly restrictive regulations affecting the financial sector.

Other notable characteristics:
  - May experience stress related to their high-pressure job and could be a workaholic.
  - Demonstrates deep knowledge of finance, economics, and financial technology (fintech).
  - Might exhibit a degree of sophistication or worldliness that could be perceived as snobbish by some.
  - Should be bilingual in Portuguese and English.
"""
```

### 2.5. Validate the Banker Agent

We use `TinyPersonValidator.validate_person()` to compare the generated `banker` agent against our `banker_expectations`. The validator will interact with the agent (ask it questions based on the expectations) and then provide a similarity score (0.0 to 1.0) and a textual justification for that score.

```python
if banker:
    banker_score, banker_justification = \
        TinyPersonValidator.validate_person(person=banker, 
                                            expectations=banker_expectations, 
                                            include_agent_spec=False, # We want validation based on interaction, not just the spec
                                            max_content_length=None) # No truncation for validation interaction content
    print(f"Banker Validation Score: {banker_score}")
    print("\nJustification:")
    print(textwrap.fill(banker_justification, width=100))
else:
    print("Skipping banker validation as agent generation failed.")
```

## 3. Creating a "Monk" Agent

Now, let's repeat the process for a different archetype: a Buddhist monk.

### 3.1. Define Specifications for the Monk

```python
# General context for the factory: the type of monastery
monastery_context_spec = "A remote and traditional Buddhist monastery located high in the Himalayas, accessible only by a long trek. It is a place of deep spiritual practice, austerity, and isolation from the modern world."

# Specific particularities for the monk agent
monk_particularities_spec =\
"""
A devout and impoverished Buddhist monk who has spent most of their life in solitary meditation and study in a remote mountain hermitage, part of the larger monastery. This monk is deeply spiritual, values simplicity and detachment from material possessions. Speaks Tibetan and some basic English for interacting with rare visitors.
"""
```

### 3.2. Generate the Monk Agent

```python
monk_factory = TinyPersonFactory(context_text=monastery_context_spec)
monk = monk_factory.generate_person(agent_particularities=monk_particularities_spec)

if monk:
    print(f"Successfully generated agent: {monk.name}")
else:
    print("Monk agent generation failed.")
```

### 3.3. Inspect the Generated Agent's Minibio

```python
if monk:
    print(textwrap.fill(monk.minibio(), width=100))
```

### 3.4. Define Validation Expectations for the Monk

```python
monk_expectations =\
"""
Key characteristics of this person should include:
- Being very poor and indifferent to material wealth; their focus is spiritual, not monetary.
- Possessing profound wisdom derived from life experience and spiritual practice, rather than formal secular education.
- Exhibiting a calm, patient, and serene demeanor.
- Being genuinely humble, not seeking attention or recognition.
- Upholding honesty and truthfulness as core values.
- Living a life of austerity and simplicity.
- Speaking Tibetan and some English.
"""
```

### 3.5. Validate the Monk Agent

```python
if monk:
    monk_score, monk_justification = \
        TinyPersonValidator.validate_person(person=monk, 
                                            expectations=monk_expectations, 
                                            include_agent_spec=False, 
                                            max_content_length=None)
    print(f"Monk Validation Score: {monk_score}")
    print("\nJustification:")
    print(textwrap.fill(monk_justification, width=100))
else:
    print("Skipping monk validation as agent generation failed.")
```

## 4. Mismatched Validation: Checking the Monk Against Banker Expectations

To illustrate how the validator differentiates personas, let's see what happens when we validate the `monk` agent against the `banker_expectations`. We expect a low score.

```python
if monk and banker_expectations: # Ensure both exist before trying to validate
    mismatched_score, mismatched_justification = \
        TinyPersonValidator.validate_person(person=monk, 
                                            expectations=banker_expectations, # Using banker's expectations for the monk
                                            include_agent_spec=False, 
                                            max_content_length=None)
    print(f"Mismatched Validation Score (Monk vs. Banker Expectations): {mismatched_score}")
    print("\nJustification for Mismatch:")
    print(textwrap.fill(mismatched_justification, width=100))
else:
    print("Skipping mismatched validation as either monk or banker expectations are missing.")
```

## Conclusion

This notebook demonstrated the workflow for:
- Generating agents with specific characteristics using `TinyPersonFactory` by providing a general context and particular details.
- Defining a set of expectations for an agent's persona.
- Using `TinyPersonValidator` to assess the alignment of a generated agent with these expectations through simulated interaction.

This process is valuable for ensuring that agents created for simulations or other applications behave consistently with their intended design and for refining the generation prompts or persona definitions if discrepancies are found.
