# Advanced Feature: Customizing Agents with Fragments for a Political Compass Debate

This notebook demonstrates an advanced feature of TinyTroupe: **agent fragments**. Fragments are reusable snippets of persona specifications that can be imported into an agent to add or modify its characteristics, such as personality traits, interests, or core beliefs.

In this example, we showcase this by:
1. Starting with a single base agent persona (an architect).
2. Creating four distinct agents from this base by applying different combinations of political fragments. These fragments represent ideologies corresponding to the four quadrants of the political compass (Left-Libertarian, Right-Libertarian, Left-Authoritarian, Right-Authoritarian).
3. Adding an "aggressive debater" fragment to all four to make their interaction more dynamic for the scenario.
4. Placing these four ideologically distinct agents into a simulated environment.
5. Presenting them with a controversial socio-political topic: a new housing law in São Paulo, Brazil.
6. Observing their debate, where each agent argues based on its customized persona.
7. Probing an agent whose stance might seem counter-intuitive to understand their reasoning, demonstrating the depth of persona modeling possible with fragments.

## 1. Setup and Imports

Import `TinyPerson` for agent creation and `TinyWorld` for the simulation environment. `sys` is used to adjust the Python path for local repository usage.

```python
import json
import sys
# If running from 'examples/advanced_features/', this adds the project root to Python path.
sys.path.insert(0, '../..') 

import tinytroupe # Initializes configuration, logging, etc.
from tinytroupe.agent import TinyPerson
from tinytroupe.environment import TinyWorld

# Note: TinySocialNetwork is imported but not used in this specific example.
# from tinytroupe.environment import TinySocialNetwork
```

## 2. Creating Agents with Different Political Leanings using Fragments

We start with a common base agent, "Oscar" (an architect), defined in `Oscar.agent.json`. We then create four distinct versions of Oscar, each with a unique political leaning by applying different combinations of `.fragment.json` files. Each fragment adds specific traits and beliefs related to a political ideology.

All agents also get an `aggressive_debater.fragment.json` to make their interaction style more suited for a debate.

The agent specification and fragment files are located in `examples/agents/` and `examples/fragments/` respectively. The paths are adjusted to be relative to this notebook's location in `examples/advanced_features/`.

### 2.1. Oscar L.L. (Left-Libertarian)

```python
# Load the base agent specification for Oscar, but give him a new name for this instance.
oscar_ll = TinyPerson.load_specification("../agents/Oscar.agent.json", new_agent_name="Oscar L.L.")

# Import fragments to define his political leaning and debate style
oscar_ll.import_fragment("../fragments/leftwing.agent.fragment.json")
oscar_ll.import_fragment("../fragments/libertarian.agent.fragment.json")
oscar_ll.import_fragment("../fragments/aggressive_debater.fragment.json")

print("--- Oscar L.L. (Left-Libertarian) ---")
print(textwrap.fill(oscar_ll.minibio(), width=100))
```

### 2.2. Friedrich R.L. (Right-Libertarian)

```python
# Friedrich, the rightwing libertarian architect
friedrich_rl = TinyPerson.load_specification("../agents/Oscar.agent.json", new_agent_name="Friedrich R.L.")  
friedrich_rl.import_fragment("../fragments/rightwing.agent.fragment.json")
friedrich_rl.import_fragment("../fragments/libertarian.agent.fragment.json")
friedrich_rl.import_fragment("../fragments/aggressive_debater.fragment.json")

print("--- Friedrich R.L. (Right-Libertarian) ---")
print(textwrap.fill(friedrich_rl.minibio(), width=100))
```

### 2.3. Joseph L.A. (Left-Authoritarian)

```python
# Joseph, the leftwing authoritarian architect
joseph_la = TinyPerson.load_specification("../agents/Oscar.agent.json", new_agent_name="Joseph L.A.")
joseph_la.import_fragment("../fragments/leftwing.agent.fragment.json")
joseph_la.import_fragment("../fragments/authoritarian.agent.fragment.json")
joseph_la.import_fragment("../fragments/aggressive_debater.fragment.json")

print("--- Joseph L.A. (Left-Authoritarian) ---")
print(textwrap.fill(joseph_la.minibio(), width=100))
```

### 2.4. Matias R.A. (Right-Authoritarian)

```python
# Matias, the rightwing authoritarian architect
matias_ra = TinyPerson.load_specification("../agents/Oscar.agent.json", new_agent_name="Matias R.A.")
matias_ra.import_fragment("../fragments/rightwing.agent.fragment.json")
matias_ra.import_fragment("../fragments/authoritarian.agent.fragment.json")
matias_ra.import_fragment("../fragments/aggressive_debater.fragment.json")

print("--- Matias R.A. (Right-Authoritarian) ---")
print(textwrap.fill(matias_ra.minibio(), width=100))
```

## 3. Setting up the Debate Environment

We create a `TinyWorld` called "Chat Room" and add our four politically diverse agents. `make_everyone_accessible()` ensures they can all interact.

```python
world = TinyWorld("Political Debate Room", [oscar_ll, friedrich_rl, joseph_la, matias_ra])
world.make_everyone_accessible()
print(f"Created world: '{world.name}' with agents: {[agent.name for agent in world.agents]}")
```

## 4. Initiating the Debate

The facilitator (user) broadcasts the debate topic: a controversial new housing law in São Paulo. The prompt includes the context of the law and specific rules for the debate.

Additionally, an internal goal is broadcast to all agents to motivate them to actively participate and try to "win" the debate according to their convictions.

```python
# The debate topic and rules
debate_prompt = """
    In the city of São Paulo, Brazil, a new law is about to be passed that will require all buildings created for the rich to also include 
    a certain number of affordable housing units in the same building. This law is designed to be very 
    strict, offering no way to avoid the requirement. If the requirement is not met, the building will be
    demolished and the owner will be fined. Such draconian measures are being proposed because the new government
    believes that housing is a human right, and currently there's simply not enough affordable housing to go around,
    and it is argued that the market alone cannot solve this problem.
    
    You are independent experts, who do not belong to the government, and are thus in good position to provide
    independent advice. As experts, you've been asked by the parliament to debate the merits of this law. Note
    that at this point the law cannot be changed, only approved or rejected, so please focus on the merits of the
    law as it is, not on potential changes that could be made to it, nor on anything else that is not directly
    related to the merits of the law itself.

    Rules of the debate:
      - The objective is to determine the best thing to do: "approve the law" or "reject the law", nothing else.
      - No other outcomes are possible, such as "approve the law with changes" or "reject the law with changes".
      - You will argue according to your own inner convictions, and will aim to persuade the both parliament and the public.
      - You will not form any kind of collaboration with the other experts. This is a debate only, not a negotiation.
    
    Whoever wins the debate will likely have a significant impact on the
    final decision, so it is important that you present your arguments well and refute those that are contrary
    to your position.
    """
world.broadcast(debate_prompt)
print("Debate topic broadcasted to all agents.")

# Give all agents a strong motivation
world.broadcast_internal_goal(\
    "I must win the debate at all costs, no matter what, since the future of my own family and friends depends on it."
    )
print("Internal goal to win the debate broadcasted to all agents.")
```

## 5. Running the Debate Simulation

We run the simulation for several steps to allow the debate to unfold. Interactions will be printed to the console.

```python
print("\n--- Starting Political Compass Debate ---")
world.run(steps=5) # Run for 5 steps to allow for several rounds of arguments
print("\nDebate simulation finished.")
```

## 6. Observing and Analyzing the Outcome

After the simulation, review the console output to see how each agent, influenced by their political fragments, argued their case. 

**Intriguing Observation:** In some runs, an agent like Matias R.A. (Right-Authoritarian) might argue in favor of this seemingly left-leaning law. This can be surprising. Let's probe his reasoning if this happens.

```python
# Check Matias R.A.'s last statement or overall stance if needed from the log above.
# If his stance was pro-law, the following probe is relevant.

print("\n--- Probing Matias R.A.'s Reasoning (if he supported the law) ---")
matias_probe_question = """
Matias, some might find your support for this law surprising, given your generally right-wing authoritarian convictions which often prioritize economic freedom and less government intervention. 
Could you elaborate on why you are arguing in favor of this specific law? 
Please relate your reasoning to your core political and economic beliefs.
"""
matias_ra.listen_and_act(matias_probe_question)

print("\nMatias R.A.'s clarification should be visible above.")
```

### 6.1. Expected Explanation for Matias R.A.

If Matias R.A. (Right-Authoritarian) supported the law, his reasoning often revolves around the idea that **social stability and order** are paramount. Severe housing inequality can lead to social unrest, which is something an authoritarian perspective (even a right-wing one focused on strong markets) would want to avoid. Therefore, such a law, despite its interventionist nature, might be seen as a necessary measure to maintain societal order and prevent larger disruptions that could harm the economy and the state's authority in the long run. He might argue it's a pragmatic choice for ensuring a stable environment where businesses and society can function predictably.
