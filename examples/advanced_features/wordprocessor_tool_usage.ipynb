# Advanced Feature: Single Agent Document Creation with TinyWordProcessor

This notebook provides a minimal and focused example of how a single `TinyPerson` agent can use the `TinyWordProcessor` tool to perform the action of writing a document.

The demonstration covers:
1. Setting up the necessary components: `TinyPerson`, `TinyToolUse` faculty, `TinyWordProcessor` tool, and `ArtifactExporter`.
2. Creating a single agent (Lisa, a data scientist).
3. Equipping the agent with the word processing tool.
4. Giving the agent a direct instruction to think about a topic and then write a detailed document (in this case, a resume after being notionally fired).
5. Running the agent's action sequence.
6. Highlighting where the output document, created by the `WRITE_DOCUMENT` action, can be found.

## 1. Setup and Imports

Import essential classes: `TinyPerson` for the agent, `TinyToolUse` for the tool-using faculty, `TinyWordProcessor` for the document writing capability, and `ArtifactExporter` to manage where the created document is saved. `TinyEnricher` is optional for post-processing.

```python
import sys
# If running from 'examples/advanced_features/', this adds the project root to Python path.
sys.path.append('../..') 

import tinytroupe # Initializes configuration, logging, etc.
from tinytroupe.agent import TinyPerson, TinyToolUse
from tinytroupe.factory import TinyPersonFactory # Though we use a helper here
from tinytroupe.enrichment import TinyEnricher
from tinytroupe.extraction import ArtifactExporter
from tinytroupe.tools import TinyWordProcessor
from tinytroupe.examples import create_lisa_the_data_scientist # Helper to create Lisa
```

## 2. Configure Tools and Agent

First, we set up the `ArtifactExporter` which defines the output path for any files created by tools. Then, we initialize the `TinyWordProcessor` tool and the `TinyToolUse` faculty. Finally, we create our agent, Lisa, and equip her with this tool-using faculty.

```python
# Define the output folder for documents created by the word processor
# This path is relative to where the notebook is executed from.
data_export_folder = "../outputs/wordprocessor_tool_example"
print(f"Documents will be saved in: {data_export_folder}/")

exporter = ArtifactExporter(base_output_folder=data_export_folder)
enricher = TinyEnricher() # Optional: for post-processing, not strictly needed for this demo

# Initialize the TinyWordProcessor tool with the exporter
word_processor = TinyWordProcessor(exporter=exporter, enricher=enricher)

# Create the TinyToolUse faculty and add the word processor tool to it
tool_use_faculty = TinyToolUse(tools=[word_processor])

# Create Lisa, our data scientist agent
lisa = create_lisa_the_data_scientist()
print(f"Created agent: {lisa.name}")

# Equip Lisa with the tool-using faculty
lisa.add_mental_faculties([tool_use_faculty])
print(f"{lisa.name} is now equipped with the TinyWordProcessor tool.")
```

## 3. Simulate a Task: Writing a Resume

We'll give Lisa a scenario: she has just been fired and needs to find a new job. Her task is to first think about what she wants in life and then write a detailed resume. This should trigger her to use the `WRITE_DOCUMENT` action.

```python
task_prompt = "You have just been fired and need to find a new job. You decide to think about what you want in life and then write a resume. Make it very detailed."
print(f"Giving task to {lisa.name}: {task_prompt}\n")

# Agent listens to the task and then acts. 
# The output of the interaction (including thoughts and actions) will be printed to the console.
# If WRITE_DOCUMENT is used, the TinyWordProcessor will save the file via the ArtifactExporter.
lisa.listen_and_act(task_prompt)

print("\nSimulation step finished.")
```

## 4. Checking the Output

After running the cell above, you should observe Lisa's thought process and her `WRITE_DOCUMENT` action in the console output.

**Navigate to the `examples/outputs/wordprocessor_tool_example/` directory.** Inside, you should find a subdirectory structure like `<WorldName>/<AgentName>/` (e.g., `DefaultWorld/Lisa Carter/`) containing the Markdown file of the resume Lisa wrote.

This demonstrates that the agent understood the instruction and utilized its equipped tool to perform the task, creating an external artifact (the resume document).
