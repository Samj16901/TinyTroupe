# Simple Chat: Basic Agent Interaction

This notebook demonstrates a fundamental example of how to use TinyTroupe to simulate a conversation between two agents, Lisa and Oscar. 
It covers:
- Loading pre-defined agent specifications.
- Creating a world for the agents.
- Initiating an interaction.
- Running the simulation for a few steps.
- Displaying the interaction history from different perspectives.

## 1. Setup and Imports

First, we import the necessary modules from the TinyTroupe library. 
- `TinyPerson` represents our agents.
- `TinyWorld` is the environment where agents interact.
- We also import `json` for loading agent data and `sys` to adjust the path for finding the `tinytroupe` library if you are running this notebook from within the `examples` directory of the repository.

```python
import json
import sys
# If running from the 'examples' directory, this adds the parent directory (project root) to the Python path
# This allows Python to find the 'tinytroupe' module.
sys.path.insert(0, '..')

import tinytroupe # Initializes configuration, logging, etc.
from tinytroupe.agent import TinyPerson
from tinytroupe.environment import TinyWorld

# The tinytroupe.examples import is typically not needed for your own simulations
# but might be used in some internal examples. We'll comment it out here for clarity in a basic example.
# from tinytroupe.examples import *
```

## 2. Load Agent Specifications

We load two agents, Lisa and Oscar, from their pre-defined JSON specification files. These files contain their personas, including names, occupations, interests, and potentially initial goals or personality traits. Using `.agent.json` files is a convenient way to manage and reuse agent definitions.

The paths are now relative to the `examples/core_concepts/` directory, so we use `../agents/` to go up one level and then into the `agents` folder.

```python
# Load Lisa, the data scientist
lisa = TinyPerson.load_specification("../agents/Lisa.agent.json")
# Load Oscar, the architect
oscar = TinyPerson.load_specification("../agents/Oscar.agent.json")

print(f"Loaded agent: {lisa.name}")
print(f"Loaded agent: {oscar.name}")
```

## 3. Create the World and Add Agents

A `TinyWorld` instance provides the environment for our agents. We give it a name and then add our loaded agents to it. `make_everyone_accessible()` ensures that all agents within this world can perceive and interact with each other.

```python
# Create a world named "Chat Room"
world = TinyWorld("Chat Room")

# Add Lisa and Oscar to the world
world.add_agents([lisa, oscar])

# Make agents in the world accessible to each other
world.make_everyone_accessible()

print(f"Created world: '{world.name}' with agents: {[agent.name for agent in world.agents]}")
```

## 4. Initiate Interaction and Run Simulation

To start the conversation, we give Lisa an initial stimulus: a thought or instruction to talk to Oscar. Then, we run the simulation for a few steps. The `world.run(steps)` method will handle the turn-by-turn interactions. TinyTroupe will print the interactions to the console by default.

```python
# Give Lisa an initial stimulus to talk to Oscar
lisa.listen("Talk to Oscar to know more about him")

# Run the simulation for 4 steps
# The interactions will be printed to the console by default 
# (TinyWorld.communication_display and TinyPerson.communication_display are True by default)
world.run(steps=4)

print("\nSimulation finished.")
```

## 5. Review Interaction Logs

After the simulation, or even during, you can inspect the interaction history. Each agent maintains its own episodic memory, which records the stimuli it perceived and the actions it took. The `pp_current_interactions()` method provides a pretty-printed view of this memory.
The world object can also display the combined interactions of all agents within it.

```python
print("\n--- Lisa's Perspective ---")
lisa.pp_current_interactions()
```

```python
print("\n--- Oscar's Perspective ---")
oscar.pp_current_interactions()
```

```python
print("\n--- World's Combined Log (All Agent Interactions) ---")
world.pp_current_interactions(max_content_length=120) # Truncate long content for brevity
```

This basic example shows the core workflow: define agents, place them in a world, provide an initial stimulus, and let the simulation unfold. Explore other examples to see more advanced features like agent generation, memory retrieval, and using tools.
